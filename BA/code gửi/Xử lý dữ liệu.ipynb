# Một số thư viện Python

Trong Python, Pandas và NumPy là hai thư viện cốt lõi được sử dụng rộng rãi trong xử lý dữ liệu, phân tích dữ liệu, và tính toán khoa học.
 ## Vai trò của NumPy
NumPy (Numerical Python) là thư viện nền tảng cho tính toán số học và xử lý mảng đa chiều trong Python. Nó cung cấp các công cụ mạnh mẽ để làm việc với dữ liệu số, đặc biệt trong các ứng dụng khoa học và kỹ thuật.

Mảng đa chiều (ndarray): NumPy cung cấp đối tượng mảng (ndarray) hiệu quả để lưu trữ và thao tác dữ liệu số (số nguyên, số thực, v.v.). Mảng này nhanh hơn nhiều so với danh sách Python nhờ tối ưu hóa ở cấp thấp.

Tính toán số học: Hỗ trợ các phép toán ma trận, đại số tuyến tính, thống kê cơ bản (trung bình, phương sai, v.v.), và các phép toán element-wise (theo từng phần tử).

Hiệu suất cao: Được viết bằng C, NumPy rất nhanh và tối ưu cho các phép toán trên tập dữ liệu lớn.

Hỗ trợ toán học: Cung cấp các hàm toán học như sin, cos, log, exp, và các công cụ cho phép tạo số ngẫu nhiên, biến đổi Fourier, v.v.

Nền tảng cho các thư viện khác: NumPy là cơ sở cho nhiều thư viện khác như Pandas, SciPy, TensorFlow, và Scikit-learn.

 Ứng dụng:
 
Xử lý dữ liệu số lớn (ví dụ: ma trận trong học máy).
Tính toán khoa học (vật lý, kỹ thuật).
Xử lý hình ảnh, tín hiệu, và dữ liệu đa chiều.

# 1. Xử lý dữ liệu với Python

```python
import numpy as np
from scipy import stats

# Dữ liệu số giờ học mỗi tuần của 10 sinh viên
data = [15, 12, 17, 10, 14, 16, 13, 18, 11, 15]

# Tính các giá trị thống kê mẫu
n = len(data)
mean = np.mean(data)
std = np.std(data, ddof=1)  # ddof=1 để tính độ lệch chuẩn mẫu

# Mức ý nghĩa alpha = 0.05 → alpha/2 = 0.025
alpha = 0.05
t_crit = stats.t.ppf(1 - alpha/2, df=n - 1)

# Tính sai số chuẩn và khoảng tin cậy
se = std / np.sqrt(n)
margin_of_error = t_crit * se
ci_lower = mean - margin_of_error
ci_upper = mean + margin_of_error

# In kết quả
print(f"Trung bình mẫu: {mean:.2f}")
print(f"Độ lệch chuẩn mẫu: {std:.2f}")
print(f"Khoảng tin cậy 95%: [{ci_lower:.2f}, {ci_upper:.2f}]")
```

```python
# ước lượng tỷ lệ cho tổng thể
import math
from scipy.stats import norm

# Dữ liệu đầu vào
n = 200          # cỡ mẫu
x = 120          # số người hài lòng
alpha = 0.05     # mức ý nghĩa

# Tính tỷ lệ mẫu
p_hat = x / n

# Giá trị tới hạn z (tra bảng Z với alpha/2)
z_crit = norm.ppf(1 - alpha/2)  # ví dụ: z0.025 ≈ 1.96 cho 95% CI

# Tính sai số chuẩn
se = math.sqrt(p_hat * (1 - p_hat) / n)

# Khoảng tin cậy
margin_of_error = z_crit * se
ci_lower = p_hat - margin_of_error
ci_upper = p_hat + margin_of_error

# In kết quả
print(f"Tỷ lệ mẫu (p̂): {p_hat:.4f}")
print(f"Giá trị z tới hạn: {z_crit:.3f}")
print(f"Sai số chuẩn: {se:.4f}")
print(f"Khoảng tin cậy 95%: [{ci_lower:.4f}, {ci_upper:.4f}]")
```

```python
# Một nhà quản lý marketing cho rằng ít nhất 70% khách hàng hài lòng với dịch vụ của công ty. Để kiểm chứng điều này, một khảo sát ngẫu nhiên được thực hiện với 150 khách hàng, trong đó có 99 người bày tỏ sự hài lòng.
#Tiến hành kiểm định giả thuyết về tỷ lệ khách hàng hài lòng với:

#Giả thuyết không (H₀): 
#𝑝≥0.70
#p≥0.70

#Giả thuyết đối (H₁): 
#𝑝<0.70
#p<0.70

#→ Đây là kiểm định một phía (trái) vì chúng ta nghi ngờ rằng tỷ lệ thực tế thấp hơn 70%.
import math
from scipy.stats import norm

# Dữ liệu đầu vào
n = 150       # kích thước mẫu
x = 99        # số người hài lòng
p0 = 0.70     # tỷ lệ theo giả thuyết

# Tính toán
p_hat = x / n
se = math.sqrt(p0 * (1 - p0) / n)
z_stat = (p_hat - p0) / se
p_value = norm.cdf(z_stat)  # kiểm định 1 phía trái

# In kết quả
print(f"Tỷ lệ mẫu (p̂): {p_hat:.4f}")
print(f"Giá trị z: {z_stat:.4f}")
print(f"p-value (1 phía): {p_value:.4f}")

# Kết luận
alpha = 0.05
if p_value < alpha:
    print("Kết luận: Bác bỏ H0 – Có bằng chứng rằng tỷ lệ hài lòng < 70%")
else:
    print("Kết luận: Không bác bỏ H0 – Chưa đủ bằng chứng rằng tỷ lệ hài lòng < 70%")
```

```python
from scipy.stats import t
import math

# Dữ liệu mẫu
sample_mean = 4.6
sample_std = 0.8
n = 20
mu_0 = 5
alpha = 0.05

# Tính thống kê kiểm định
se = sample_std / math.sqrt(n)
t_stat = (sample_mean - mu_0) / se

# Tính p-value (1 phía trái)
p_value = t.cdf(t_stat, df=n - 1)

# In kết quả
print(f"T-statistic: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")

# Kết luận
if p_value < alpha:
    print("✅ Bác bỏ H₀: Thời gian xử lý hồ sơ đã giảm.")
else:
    print("⚠️ Không bác bỏ H₀: Chưa đủ bằng chứng rằng thời gian xử lý đã giảm.")
```

# Xử lý outliers

```python
import matplotlib.pyplot as plt # Plotting graphs
import seaborn as sns
# Identify outliers using IQR method
def identify_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers

# Example: Identify outliers in GDP
Total_Trans_Amt_outliers = identify_outliers(df, 'Total_Trans_Amt')

print("Number of outliers in Total_Trans_Amt:", len(Total_Trans_Amt_outliers))
print("\
Top 5 Total_Trans_Amt outliers:")
print(Total_Trans_Amt_outliers[['CLIENTNUM', 'Total_Trans_Amt']].sort_values('Total_Trans_Amt', ascending=False).head())

# Visualize Total_Trans_Amt distribution
plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Total_Trans_Amt'])
plt.title('Total_Trans_Amt Distribution')
plt.xlabel('Total_Trans_Amt')
plt.show()
```

```python
# Handle outliers using capping (winsorization)
import numpy as np
csv = pd.read_csv("Bankchurners.csv")
def cap_outliers(cvs, column):
    Q1 = csv[column].quantile(0.25)
    Q3 = csv[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    csv[column] = np.clip(cvs[column], lower_bound, upper_bound) 
    return csv
    #Hàm clip() trong NumPy dùng để giới hạn giá trị: ép mọi giá trị nhỏ hơn lower_bound thành lower_bound, và mọi giá trị lớn hơn upper_bound thành upper_bound
    


numerical_columns = csv.select_dtypes(include=[np.number]).columns

# Cap outliers for all numerical columns
for col in numerical_columns:
    csv = cap_outliers(csv, col)   # cap_outliers dùng để tự giới hạn giá trị ngoại lai trên cột col

# Scale numerical features
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
csv[numerical_columns] = scaler.fit_transform(csv[numerical_columns])
```

```python
# Display summary statistics after preprocessing
print("Summary statistics after preprocessing:")
print(csv[numerical_columns].describe())

# Save preprocessed data
csv.to_csv('preprocessed_countries.csv', index=False)  # xuất ra file dữ liệu preprocessed_countries.csv
print("\
Preprocessed data saved to 'preprocessed_countries.csv'")
```
