## 🛠️ 1. Khởi tạo môi trường & Đọc dữ liệu gốc `customer.csv`

```python
import pandas as pd
import numpy as np
import os

# Cấu hình hiển thị
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', lambda x: '%.4f' % x)

# Đọc dữ liệu gốc
data_path = 'customer.csv' if os.path.exists('customer.csv') else 'BA/code gửi/customer.csv'
df = pd.read_csv(data_path)

print(f"✅ Tải dữ liệu gốc thành công: {df.shape[0]:,} dòng, {df.shape[1]} cột.")
df.head()
```

---
## 📌 CÂU 1: Trích xuất dữ liệu gồm các biến định tính và định lượng
> **Yêu cầu:** Phân tách và trích xuất các tập dữ liệu biến định tính (Qualitative) và biến định lượng (Quantitative), đồng thời xuất ra các file dữ liệu tương ứng.

```python
# 1. Xác định danh sách các cột theo kiểu dữ liệu
quant_cols = df.select_dtypes(include=[np.number]).columns.tolist()
qual_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

print(f"• Số biến định lượng ({len(quant_cols)} biến): {quant_cols}")
print(f"• Số biến định tính ({len(qual_cols)} biến): {qual_cols}")

# 2. Trích xuất DataFrame định lượng (kèm mã khách hàng Client_Num để đối soát)
df_quantitative = df[quant_cols].copy()

# 3. Trích xuất DataFrame định tính (kèm mã khách hàng Client_Num)
df_qualitative = df[['Client_Num'] + qual_cols].copy()

# 4. Xuất file dữ liệu CSV và Excel
df_quantitative.to_csv('customer_quantitative.csv', index=False)
df_qualitative.to_csv('customer_qualitative.csv', index=False)

print("\n📁 Đã xuất 2 file thành công:")
print("  1. customer_quantitative.csv (Dữ liệu định lượng)")
print("  2. customer_qualitative.csv (Dữ liệu định tính)")

print("\n=== MẪU DỮ LIỆU ĐỊNH LƯỢNG ===")
display(df_quantitative.head())

print("\n=== MẪU DỮ LIỆU ĐỊNH TÍNH ===")
display(df_qualitative.head())
```

---
## 📌 CÂU 2: Trích xuất dữ liệu khách hàng ĐÃ KẾT HÔN (`Marital_Status == 'Married'`)
> **Yêu cầu:** Lấy trích xuất dữ liệu thông tin với các khách hàng có tình trạng hôn nhân là 'Married'.

```python
# Lọc khách hàng có Marital_Status là 'Married'
df_married = df[df['Marital_Status'] == 'Married'].copy()

count_married = len(df_married)
pct_married = (count_married / len(df)) * 100

print(f"💍 Số lượng khách hàng ĐÃ KẾT HÔN: {count_married:,} khách ({pct_married:.2f}% tổng số khách hàng)")

# Xuất file dữ liệu
df_married.to_csv('customer_married.csv', index=False)
print("📁 Đã lưu file: customer_married.csv")

display(df_married.head())
```

---
## 📌 CÂU 3: Trích xuất dữ liệu khách hàng CÓ SỞ HỮU NHÀ (`House_Owner == 'yes'`)
> **Yêu cầu:** Lấy ra thông tin các khách hàng có sở hữu nhà ở.

```python
# Lọc khách hàng có House_Owner là 'yes'
df_house_owner = df[df['House_Owner'].str.lower() == 'yes'].copy()

count_house = len(df_house_owner)
pct_house = (count_house / len(df)) * 100

print(f"🏠 Số lượng khách hàng SỞ HỮU NHÀ: {count_house:,} khách ({pct_house:.2f}% tổng số khách hàng)")

# Xuất file dữ liệu
df_house_owner.to_csv('customer_house_owners.csv', index=False)
print("📁 Đã lưu file: customer_house_owners.csv")

display(df_house_owner.head())
```

---
## 📌 CÂU 4: Trích xuất dữ liệu khách hàng ở BANG CALIFORNIA (`state_cd == 'CA'`)
> **Yêu cầu:** Lấy thông tin các khách hàng cư trú tại bang California (mã bang 'CA').

```python
# Lọc khách hàng ở bang CA
df_ca = df[df['state_cd'] == 'CA'].copy()

count_ca = len(df_ca)
pct_ca = (count_ca / len(df)) * 100

print(f"🌴 Số lượng khách hàng ở BANG CALIFORNIA (CA): {count_ca:,} khách ({pct_ca:.2f}% tổng số khách hàng)")

# Xuất file dữ liệu
df_ca.to_csv('customer_california.csv', index=False)
print("📁 Đã lưu file: customer_california.csv")

display(df_ca.head())
```

---
## 📌 CÂU 5: Trích xuất dữ liệu khách hàng TRONG ĐỘ TUỔI 54 - 64
> **Yêu cầu:** Lấy thông tin các khách hàng nằm trong độ tuổi từ 54 đến 64 tuổi (`54 <= Customer_Age <= 64`).

```python
# Lọc khách hàng có độ tuổi từ 54 đến 64 (inclusive)
df_age_54_64 = df[df['Customer_Age'].between(54, 64)].copy()

count_age_54_64 = len(df_age_54_64)
pct_age_54_64 = (count_age_54_64 / len(df)) * 100

print(f"🎯 Số lượng khách hàng ĐỘ TUỔI 54-64: {count_age_54_64:,} khách ({pct_age_54_64:.2f}% tổng số khách hàng)")
print(f"   • Tuổi nhỏ nhất trong tập: {df_age_54_64['Customer_Age'].min()}")
print(f"   • Tuổi lớn nhất trong tập: {df_age_54_64['Customer_Age'].max()}")

# Xuất file dữ liệu
df_age_54_64.to_csv('customer_age_54_64.csv', index=False)
print("📁 Đã lưu file: customer_age_54_64.csv")

display(df_age_54_64.head())
```
