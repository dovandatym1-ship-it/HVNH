# Xử lý outliers

```python
# Identify outliers using IQR method
def identify_outliers(excel, column):
    Q1 = excel[column].quantile(0.25)
    Q3 = excel[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = excel[(excel[column] < lower_bound) | (excel[column] > upper_bound)]
    return outliers

# Tìm outliers trong biến EV
ev_outliers = identify_outliers(excel, 'EV')

# In số lượng và top 5 outliers
print("Number of outliers in EV:", len(ev_outliers))
print("Top 5 EV outliers:")
print(ev_outliers[['Firm', 'EV']].sort_values('EV', ascending=False).head())


plt.figure(figsize=(10, 6))
sns.boxplot(x=excel['EV'])
plt.title('EV Distribution')
plt.xlabel('EV')
plt.show()
```

```python
# Đọc dữ liệu từ Excel
excel = pd.read_excel('/content/drive/My Drive/Phan tich kinh doanh/DATA2201.xlsx')

# Hàm xử lý outliers bằng Winsorization (capping)
def cap_outliers(excel, column):
    Q1 = excel[column].quantile(0.25)
    Q3 = excel[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    excel[column] = np.clip(excel[column], lower_bound, upper_bound)  # ép giá trị về trong khoảng hợp lý
    return excel

# Lấy danh sách các cột dạng số (numerical)
numerical_columns = excel.select_dtypes(include=[np.number]).columns

# Xử lý outliers bằng capping cho từng cột số
for col in numerical_columns:
    excel = cap_outliers(excel, col)

# Chuẩn hóa các biến số với StandardScaler
scaler = StandardScaler()
excel[numerical_columns] = scaler.fit_transform(excel[numerical_columns])

```

```python
# Display summary statistics after preprocessing
print("Summary statistics after preprocessing:")
print(excel[numerical_columns].describe())

# Save preprocessed data
excel.to_excel('/content/drive/My Drive/Phan tich kinh doanh/processedDATA2201.xlsx', index=False)
print("\Preprocessed data saved to 'processedDATA2201.xlsx'")

```

# Correlation Matrix and Heat Map

Hệ số tương quan là một chỉ số thống kê đo lường mức độ và chiều hướng mối quan hệ tuyến tính giữa hai biến số, nhận giá trị trên [-1, 1].
Công thức tính hệ số tương quan Pearson

 $r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2} \cdot \sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}}$

$x_i, y_i$: các giá trị quan sát của hai biến X và Y

$\bar x, \bar y$: giá trị trung bình của X và Y

Nếu r = 1: Tương quan tuyến tính dương hoàn hảo

0 < r < 1: Tương quan dương cùng nhiều

r = 0: không có tương quan dương tuyến tính

-1 < r < 0: Tương quan âm nghịch chiều

r = -1: Tương quan tuyến tính âm


```python
# Tính hệ số tương quan của các biến nhận giá trị numeric
df = pd.read_excel('/content/drive/My Drive/Phan tich kinh doanh/processedDATA2201.xlsx')
numeric_columns = excel.select_dtypes(include=[np.number]).columns  # lấy các cột có giá trị là numeric
corr = excel[numeric_columns].corr('pearson')
```

```python
plt.figure(figsize = (15,15))
sns.heatmap(corr, vmax=1.0, vmin=-1.0, square=True, annot=True, annot_kws={"size": 9, "color": "black"},
            linewidths=0.1, cmap='rocket')
```

DecisionTree

```python
# 1. Thư viện cần thiết
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import numpy as np

# 2. Đọc dữ liệu
df = pd.read_excel('/content/drive/My Drive/Phan tich kinh doanh/processedDATA2201.xlsx')

# 3. Tách biến đầu vào và mục tiêu
X = df.iloc[:, 3:]                # từ cột thứ 4 trở đi (cột thứ 3 nếu tính từ 0)
y = df['EV']                      # biến mục tiêu

# Gộp và loại bỏ các dòng chứa NaN
data = pd.concat([X, y], axis=1)
data_clean = data.dropna()

X_clean = data_clean.drop('EV', axis=1)
y_clean = data_clean['EV']

# 4. Chia dữ liệu train - test
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)

# 5. Huấn luyện mô hình Decision Tree Regressor
model = DecisionTreeRegressor(random_state=42)
model.fit(X_train, y_train)

# 6. Dự đoán
y_pred = model.predict(X_test)

# 7. Đánh giá mô hình
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")

# 8. Vẽ biểu đồ dự báo
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.5, color='green')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('Actual EV')
plt.ylabel('Predicted EV')
plt.title('Actual vs Predicted (Decision Tree Regression)')
plt.grid()
plt.show()
```

```python
df = pd.read_excel('/content/drive/My Drive/Phan tich kinh doanh/processedDATA2201.xlsx')
# 1. Thư viện cần thiết
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import numpy as np
from sklearn.tree import DecisionTreeRegressor # Import DecisionTreeRegressor
from xgboost import XGBRegressor # Import XGBRegressor


# 2. Tách biến đầu vào và mục tiêu
X = df.iloc[:, 3:]                # từ cột thứ 6 trở đi
y = df['EV']          # biến mục tiêu là log_indexprice
data = pd.concat([X, y], axis=1)
data_clean = data.dropna()
X_clean = data_clean.drop('EV', axis=1)
y_clean = data_clean['EV']
# 3. Chia train - test
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)

# 4. Huấn luyện mô hình Random Forest Regressor
model1 = RandomForestRegressor(n_estimators=100, random_state=42)
model1.fit(X_train, y_train)
# model2 = DecisionTreeRegressor(random_state=42) # Corrected class name
# model2.fit(X_train, y_train)
# model3 = XGBRegressor(random_state=42) # Corrected class name
# model3.fit(X_train, y_train)
# 5. Dự đoán
y_pred = model1.predict(X_test) # Use model1 for prediction

# 6. Đánh giá mô hình
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")

# 7. Vẽ biểu đồ dự báo
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.5, color='blue')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('Actual log_indexprice')
plt.ylabel('Predicted log_indexprice')
plt.title('Actual vs Predicted (Random Forest Regression)')
plt.grid()
plt.show()
```

# Mô hình Xgboost

```python
# 1. Thư viện cần thiết
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from xgboost import XGBRegressor

# 2. Đọc dữ liệu
df = pd.read_excel('/content/drive/My Drive/Phan tich kinh doanh/processedDATA2201.xlsx')

# 3. Tách biến đầu vào và mục tiêu
X = df.iloc[:, 3:]                # từ cột thứ 4 trở đi
y = df['EV']                      # biến mục tiêu

# Gộp và loại bỏ NaN
data = pd.concat([X, y], axis=1)
data_clean = data.dropna()

X_clean = data_clean.drop('EV', axis=1)
y_clean = data_clean['EV']

# 4. Chia tập train - test
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)

# 5. Huấn luyện mô hình XGBoost Regressor
model = XGBRegressor(random_state=42, n_estimators=100)
model.fit(X_train, y_train)

# 6. Dự đoán
y_pred = model.predict(X_test)

# 7. Đánh giá mô hình
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")

# 8. Vẽ biểu đồ dự báo
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.5, color='orange')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('Actual EV')
plt.ylabel('Predicted EV')
plt.title('Actual vs Predicted (XGBoost Regression)')
plt.grid()
plt.show()
```

```python
# Tầm quan trọng của các biến
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Lấy tên các biến đầu vào
feature_names = X.columns

# Lấy giá trị độ quan trọng từ mô hình
importances = model.feature_importances_

# Tạo DataFrame sắp xếp theo độ quan trọng giảm dần
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

# In bảng
print(importance_df)

# Vẽ biểu đồ
plt.figure(figsize=(10,6))
plt.barh(importance_df['Feature'], importance_df['Importance'], color='skyblue')
plt.xlabel("Feature Importance")
plt.title("Tầm quan trọng của các biến (XGBoost)")
plt.gca().invert_yaxis()  # để biến quan trọng nhất ở trên cùng
plt.grid(axis='x')
plt.tight_layout()
plt.show()
```
