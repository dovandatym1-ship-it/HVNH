```python
def identify_outliers_iqr(df, column, factor=1.5):
    """
    Trả về:
      outliers_df: DataFrame chứa các hàng là outliers theo cột `column`
      mask: Series boolean cùng index với df (True = outlier)
      lower_bound, upper_bound: giá trị biên dưới/biên trên
    factor: hệ số IQR (mặc định 1.5)
    """
    # 1) đảm bảo lấy chuỗi dạng số, nếu không chuyển được sẽ thành NaN
    series = pd.to_numeric(df[column], errors='coerce')

    # 2) tính Q1 (25%) và Q3 (75%)
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    # 3) IQR = Q3 - Q1
    IQR = Q3 - Q1

    # 4) tính biên dưới và biên trên theo quy tắc 1.5*IQR (hoặc factor*IQR)
    lower_bound = Q1 - factor * IQR
    upper_bound = Q3 + factor * IQR

    # 5) tạo mask (True nếu < lower_bound hoặc > upper_bound)
    mask = (series < lower_bound) | (series > upper_bound)

    # 6) lấy các hàng là outliers (copy để an toàn)
    outliers_df = df[mask].copy()

    # trả về outliers, mask để dùng tiếp, và biên
    return outliers_df, mask, lower_bound, upper_bound
```

```python
df = pd.read_excel("DATA2201.xlsx")
```

```python
def cap_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[column] = np.clip(df[column], lower_bound, upper_bound)
    return df
```

```python
numerical_columns = df.select_dtypes(include=[np.number]).columns
```

```python
for col in numerical_columns:
    df = cap_outliers(df, col)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df[numerical_columns] = scaler.fit_transform(df[numerical_columns])
```

```python
# Tính hệ số tương quan Pearson cho tất cả biến số
corr = df.select_dtypes(include=[np.number]).corr(method='pearson')

# Lấy tương quan của EV với các biến khác
corr_ev = corr['EV'].drop(['EV']).sort_values(ascending=False)

print("Tương quan của EV với các biến khác:")
print(corr_ev)
```

```python
# Chọn các biến tương quan mạnh với EV, ngưỡng 0.5 thay vì 0.6
strong_corr = corr_ev[abs(corr_ev) >= 0.5].sort_values(ascending=False).index.tolist()
print("Strongly correlated features with EV (threshold 0.5):", strong_corr)

# Tạo DataFrame chứa EV + các biến này
strong_fet = df.loc[:, strong_corr + ['EV']]

# Vẽ plot cho tất cả biến tương quan
fig, axes = plt.subplots(1, len(strong_corr), figsize=(16, 6))  
if len(strong_corr) == 1:  # nếu chỉ 1 plot thì axes không phải list
    axes = [axes]

for i, col in enumerate(strong_corr):
    sns.regplot(
        x=col,
        y='EV',
        data=strong_fet,
        ax=axes[i],
        line_kws={'color': 'red'}
    )
    axes[i].set_title(f'{col} vs EV')

plt.tight_layout()
plt.show()
```
