# xử lý Missing value

```python
numeric_columns = df.select_dtypes(include=[np.number]).columns  # lấy các cột có giá trị là numeric
df[numeric_columns] = df[numeric_columns].fillna(df[numeric_columns].median())   # điền missing value bởi median bởi các biến có dang numeric
```

```python
data_df = df.copy()

# df là dataframe đã đọc từ SQL
data_df['EV'] = np.log(data_df['EV'])
data_df['FS'] = np.log(data_df['FS'])
```

```python
data_df.describe()
```

```python
df = data_df.copy()
```

```python
from sklearn.preprocessing import MinMaxScaler

# 1. Chọn các cột numeric
numeric_cols = df.select_dtypes(include=['number', 'float']).columns

# 2. Thay thế giá trị vô cùng (inf, -inf) thành NaN
df[numeric_cols] = df[numeric_cols].replace([np.inf, -np.inf], np.nan)

# 3. Xử lý NaN (ví dụ: thay bằng median hoặc 0)
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# 4. Scale dữ liệu về [-1, 1]
scaler = MinMaxScaler(feature_range=(-1, 1))
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print(df.head())
```

# Correlation Matrix and Heat Map

Hệ số tương quan là một chỉ số thống kê đo lường mức độ và chiều hướng mối quan hệ tuyến tính giữa hai biến số, nhận giá trị trên [-1, 1].
Công thức tính hệ số tương quan Pearson
 
 $r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2} \cdot \sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}}$

$x_i, y_i$: các giá trị quan sát của hai biến X và Y

$\bar x, \bar y$: giá trị trung bình của X và Y

Nếu r = 1: Tương quan tuyến tính dương hoàn hảo

0 < r < 1: Tương quan dương cùng nhiều

r = 0: không có tương quan dương tuyến tính

-1 < r < 0: Tương quan âm nghịch chiều

r = -1: Tương quan tuyến tính âm 


```python
# Tính hệ số tương quan của các biến nhận giá trị numeric
numeric_columns = df.select_dtypes(include=[np.number]).columns  # lấy các cột có giá trị là numeric
corr = df[numeric_columns].corr('pearson')
```

```python
plt.figure(figsize = (15,15))
sns.heatmap(corr, vmax=1.0, vmin=-1.0, square=True, annot=True, annot_kws={"size": 9, "color": "black"}, 
            linewidths=0.1, cmap='rocket')
```

**Correlations with Index Prices**

```python
# tính hệ số tương quan của 1 biến EV với các biến còn lại
corr1 = corr['EV'].drop(['EV'])
corr1.sort_values(ascending=False)
```

# Linear Regression

```python
# Chọn những cột có |correlation| > 0.01 (ngoại trừ chính EV)
selected_features = corr1[abs(corr1) > 0.01].index
print("Selected features:", selected_features.tolist())

# Tạo biến X với các cột này
X = df[selected_features].copy()
Y = df['EV']  # target

print("Shape of X:", X.shape)
print("Shape of Y:", Y.shape)
```

```python
# Kiểm tra xem X có chứa giá trị mising không
print(X.isna().any().any())
```

```python
# Kiểm tra chi tiết missing
print(X.isna().sum())
```

```python
# chia tập dữ liệu theo ngẫu  nhiên
from sklearn.model_selection import train_test_split

# Chia dữ liệu: 75% train, 25% test
X_train1, X_test1, y_train1, y_test1 = train_test_split(
    X, Y, test_size=0.25, random_state=42
)

print("Train size:", X_train1.shape)
print("Test size:", X_test1.shape)
```

```python
# Huấn luyện mô hình hồi quy tuyến tính
model = LinearRegression()
model.fit(X_train1, y_train1)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, r2_score
# Dự đoán và đánh giá mô hình
y_pred1 = model.predict(X_test1)

mse = mean_squared_error(y_test1, y_pred1)
rmse = mean_squared_error(y_test1, y_pred1, squared=False)
r2 = r2_score(y_test1, y_pred1)

print("📊 Kết quả đánh giá mô hình:")
print("MSE:", mse)
print("RMSE:", rmse)
print("R-squared:", r2)
```

```python
# Mô hình OLS
import statsmodels.api as sm

# Thêm cột intercept (hằng số) cho mô hình OLS
X = sm.add_constant(X)
# Bước 3: Chia dữ liệu train/test
X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X, Y, test_size=0.75, random_state=42
)

# Khởi tạo mô hình OLS và fit
model = sm.OLS(y_train2, X_train2).fit()

# In kết quả tóm tắt mô hình
print(model.summary())
```

# Random forest

```python

from sklearn.ensemble import RandomForestRegressor

# 4. Chia train - test
X_train3, X_test3, y_train3, y_test3 = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# 5. Huấn luyện mô hình Random Forest Regressor
model3 = RandomForestRegressor(n_estimators=100, random_state=42)
model3.fit(X_train3, y_train3)

# 6. Dự đoán
y_pred3 = model3.predict(X_test3)
```

```python
# 7. Đánh giá
mse = mean_squared_error(y_test3, y_pred3)
rmse = np.sqrt(mse)
r2 = r2_score(y_test3, y_pred3)

print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")
```

```python
# Vẽ biểu đồ Actual vs Predicted
plt.figure(figsize=(8,6))
plt.scatter(y_test3, y_pred3, alpha=0.5, color='blue')
plt.plot([y_test3.min(), y_test3.max()], [y_test3.min(), y_test3.max()], 'k--', lw=2)
plt.xlabel('Actual EV')
plt.ylabel('Predicted EV')
plt.title('Actual vs Predicted (Random Forest Regression)')
plt.grid()
plt.show()
```

# Neural network

```python
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor

# 3. Chia train - test
X_train4, X_test4, y_train4, y_test4 = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# 4. Chuẩn hóa dữ liệu (Neural Network cần dữ liệu có phân phối đều)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train4)
X_test_scaled = scaler.transform(X_test4)

# 5. Huấn luyện mô hình Neural Network
model4 = MLPRegressor(hidden_layer_sizes=(100, 50),  # 2 hidden layers: 100 và 50 neurons
                     activation='relu',
                     solver='adam',
                     max_iter=500,
                     random_state=42)

model4.fit(X_train_scaled, y_train4)

# 6. Dự đoán
y_pred4 = model4.predict(X_test_scaled)
```

```python
# Đánh giá mô hình
mse = mean_squared_error(y_test4, y_pred4)
rmse = np.sqrt(mse)
r2 = r2_score(y_test4, y_pred4)

print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")
```

```python
# Biểu đồ thực vs dự đoán
plt.figure(figsize=(8,6))
plt.scatter(y_test4, y_pred4, alpha=0.5, color='purple')
plt.plot([y_test4.min(), y_test4.max()], [y_test4.min(), y_test4.max()], 'k--', lw=2)
plt.xlabel('Actual EV')
plt.ylabel('Predicted EV')
plt.title('Actual vs Predicted (Neural network)')
plt.grid()
plt.show()
```

# Mô hình Xgboost

```python
from xgboost import XGBRegressor
from sklearn.preprocessing import StandardScaler

# 3. Chia train - test
X_train5, X_test5, y_train5, y_test5 = train_test_split(
    X, Y, test_size=0.2, random_state=42
)


# 4. Chuẩn hóa dữ liệu
scaler = StandardScaler()
X_train_scaled2 = scaler.fit_transform(X_train5)
X_test_scaled2 = scaler.transform(X_test5)

# 5. Khởi tạo và huấn luyện mô hình XGBoost
model5 = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

model5.fit(X_train_scaled2, y_train5)

# 6. Dự báo
y_pred5 = model5.predict(X_test_scaled2)
```

```python

# 7. Đánh giá mô hình
rmse = np.sqrt(mean_squared_error(y_test5, y_pred5))
r2 = r2_score(y_test5, y_pred5)
print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")

# 8. Vẽ biểu đồ dự báo vs thực tế
plt.figure(figsize=(8,6))
plt.scatter(y_test5, y_pred5, alpha=0.5, color='teal')
plt.plot([y_test5.min(), y_test5.max()], [y_test5.min(), y_test5.max()], 'r--', lw=2)
plt.xlabel('Actual EV')
plt.ylabel('Predicted EV')
plt.title('Actual vs Predicted (XGBoost)')
plt.grid()
plt.show()
```
