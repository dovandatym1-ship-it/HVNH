# Xử lý outliers

```python
# Identify outliers using IQR method
def identify_outliers(csv, column):
    Q1 = csv[column].quantile(0.25)
    Q3 = csv[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = csv[(csv[column] < lower_bound) | (csv[column] > upper_bound)]
    return outliers

# Example: Identify outliers in GDP
gdp_outliers = identify_outliers(csv, 'gdppercent')

print("Number of outliers in GDP:", len(gdp_outliers))
print("\
Top 5 GDP outliers:")
print(gdp_outliers[['country', 'gdppercent']].sort_values('gdppercent', ascending=False).head())

# Visualize GDP distribution
plt.figure(figsize=(10, 6))
sns.boxplot(x=csv['gdppercent'])
plt.title('GDP Distribution')
plt.xlabel('GDP')
plt.show()
```

```python
# Handle outliers using capping (winsorization)
csv = pd.read_csv("Economic Data.csv")
def cap_outliers(cvs, column):
    Q1 = csv[column].quantile(0.25)
    Q3 = csv[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    csv[column] = np.clip(cvs[column], lower_bound, upper_bound)
    return csv
    #Hàm clip() trong NumPy dùng để giới hạn giá trị: ép mọi giá trị nhỏ hơn lower_bound thành lower_bound, và mọi giá trị lớn hơn upper_bound thành upper_bound


# List of numerical columns to process
#numerical_columns = ['latitude', 'longitude', 'agricultural_land', 'forest_area', 'land_area',
                    # 'rural_land', 'urban_land', 'central_government_debt_pct_gdp', 'expense_pct_gdp',
                    # 'gdp', 'inflation', 'self_employed_pct', 'tax_revenue_pct_gdp', 'unemployment_pct',
                    # 'vulnerable_employment_pct', 'electricity_access_pct', 'alternative_nuclear_energy_pct']
numerical_columns = csv.select_dtypes(include=[np.number]).columns

# Cap outliers for all numerical columns
for col in numerical_columns:
    csv = cap_outliers(csv, col)   # cap_outliers dùng để tự giới hạn giá trị ngoại lai trên cột col

# Scale numerical features
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
csv[numerical_columns] = scaler.fit_transform(csv[numerical_columns])

```

# Correlation Matrix and Heat Map

Hệ số tương quan là một chỉ số thống kê đo lường mức độ và chiều hướng mối quan hệ tuyến tính giữa hai biến số, nhận giá trị trên [-1, 1].
Công thức tính hệ số tương quan Pearson

 $r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2} \cdot \sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}}$

$x_i, y_i$: các giá trị quan sát của hai biến X và Y

$\bar x, \bar y$: giá trị trung bình của X và Y

Nếu r = 1: Tương quan tuyến tính dương hoàn hảo

0 < r < 1: Tương quan dương cùng nhiều

r = 0: không có tương quan dương tuyến tính

-1 < r < 0: Tương quan âm nghịch chiều

r = -1: Tương quan tuyến tính âm

**Regression Plots as per strengths of correlation with index prices**

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Identifying the strong correlated features (corr > 0.6)
strong_corr = corr1[abs(corr1) >= 0.6].sort_values(ascending=False).index.tolist() # các biến có tương quan mạnh với index price
print('\nStrongly correlated features:', strong_corr, '\n')

# Strong features
strong_fet = csv.loc[:, strong_corr + ['log_indexprice']]  # trích xuất các biến tương quan mạnh với index price và index price

# Plot of strong feature's regression
fig, axes = plt.subplots(2, 2, figsize=(16, 12))  # đổi tên ax → axes
axes = axes.flatten()  # chuyển về 1 chiều

for i in range(len(strong_corr)):
    sns.regplot(
        x=strong_corr[i],
        y='log_indexprice',
        data=strong_fet,
        ax=axes[i],
        line_kws={'color': 'red'}
    )
    axes[i].set_title(f'{strong_corr[i]} vs log_indexprice')

plt.tight_layout()
plt.show()
```

```python
# moderate correlation features ( > 0.35 & < 0.6)

moderate_corr = corr1[(abs(corr1) >= 0.35) & (abs(corr1) < 0.6)].sort_values(ascending=False).index.tolist()
print('\n Moderate correlation features: ', moderate_corr, '\n')

moderate_fet = csv.loc[:, moderate_corr + ['log_indexprice']]

fig, ax = plt.subplots(2, 2, figsize=(30, 30))

for i, ax in enumerate(fig.axes):
    if i < len(moderate_corr):
        sns.regplot(x=moderate_corr[i], y='log_indexprice', data=moderate_fet, ax=ax, line_kws={'color': 'red'})
```

```python
weak_corr = corr1[(abs(corr1) < 0.35)].sort_values(ascending=False).index.tolist()
print('\n Weakly correlated features: ', weak_corr, '\n')

weak_fet = csv.loc[:, weak_corr + ['log_indexprice']]

fig, ax = plt.subplots(3, 2, figsize=(30, 30))

for i, ax in enumerate(fig.axes):
    if i < len(weak_corr):
        sns.regplot(x=weak_corr[i], y='log_indexprice', data=weak_fet, ax=ax, line_kws={'color': 'red'})
```

```python
# Huấn luyện mô hình xử lý giá trị thiếu bằng thay thế giá trị trung bình
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, r2_score

# Giả sử bạn đã có DataFrame df và biến mục tiêu là 'log_indexprice'
X = df.iloc[:, 5:]  # Cột thứ 6 có chỉ số là 5 vì bắt đầu từ 0
y = df['log_indexprice']

# Bước 1: Kiểm tra NaN
print("Số lượng giá trị thiếu mỗi cột:\n", X.isna().sum())

# Bước 2: Chia train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.75, random_state=42)

# Bước 3: Xử lý NaN bằng trung bình (mean imputation)
imputer = SimpleImputer(strategy='mean')

X_train_imputed = pd.DataFrame(imputer.fit_transform(X_train), columns=X_train.columns)
X_test_imputed = pd.DataFrame(imputer.transform(X_test), columns=X_test.columns)

# Bước 4: Huấn luyện mô hình hồi quy tuyến tính
model = LinearRegression()
model.fit(X_train_imputed, y_train)

# Bước 5: Dự đoán và đánh giá mô hình
y_pred = model.predict(X_test_imputed)

mse = mean_squared_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred, squared=False)
r2 = r2_score(y_test, y_pred)

print("📊 Kết quả đánh giá mô hình:")
print("MSE:", mse)
print("RMSE:", rmse)
print("R-squared:", r2)
```

# Neural network

```python
df = pd.read_csv('preprocessed_countries.csv')
# 1. Import thư viện
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt

# 2. Tách biến đầu vào và mục tiêu
X = df.iloc[:, 5:]                # từ cột thứ 6 trở đi
y = df['log_indexprice']          # biến mục tiêu là log_indexprice
data = pd.concat([X, y], axis=1)
data_clean = data.dropna()
X_clean = data_clean.drop('log_indexprice', axis=1)
y_clean = data_clean['log_indexprice']
# 3. Chia train - test
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)

# 4. Chuẩn hóa dữ liệu (Neural Network cần dữ liệu có phân phối đều)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 5. Huấn luyện mô hình Neural Network
model = MLPRegressor(hidden_layer_sizes=(100, 50),  # 2 hidden layers: 100 và 50 neurons
                     activation='relu',
                     solver='adam',
                     max_iter=500,
                     random_state=42)

model.fit(X_train_scaled, y_train)

# 6. Dự đoán
y_pred = model.predict(X_test_scaled)

# 7. Đánh giá mô hình
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")

# 8. Biểu đồ thực vs dự đoán
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.5, color='purple')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2)
plt.xlabel('Actual log_indexprice')
plt.ylabel('Predicted log_indexprice')
plt.title('Actual vs Predicted (Neural Network)')
plt.grid()
plt.show()
```

```python
df = pd.read_csv('preprocessed_countries.csv')
# 1. Import thư viện
import pandas as pd
import numpy as np
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# 2. Tách biến đầu vào và mục tiêu
X = df.iloc[:, 5:]                # từ cột thứ 6 trở đi
y = df['log_indexprice']          # biến mục tiêu là log_indexprice
data = pd.concat([X, y], axis=1)
data_clean = data.dropna()
X_clean = data_clean.drop('log_indexprice', axis=1)
y_clean = data_clean['log_indexprice']
# 3. Chia train - test
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42
)


# 4. Chuẩn hóa dữ liệu
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 5. Khởi tạo và huấn luyện mô hình XGBoost
model = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

model.fit(X_train_scaled, y_train)

# 6. Dự báo
y_pred = model.predict(X_test_scaled)

# 7. Đánh giá mô hình
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)
print(f"RMSE: {rmse:.4f}")
print(f"R² Score: {r2:.4f}")

# 8. Vẽ biểu đồ dự báo vs thực tế
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha=0.5, color='teal')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel('Actual log_indexprice')
plt.ylabel('Predicted log_indexprice')
plt.title('Actual vs Predicted (XGBoost)')
plt.grid()
plt.show()
```

```python
# Tầm quan trọng của các biến
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Lấy tên các biến đầu vào
feature_names = X.columns

# Lấy giá trị độ quan trọng từ mô hình
importances = model.feature_importances_

# Tạo DataFrame sắp xếp theo độ quan trọng giảm dần
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
}).sort_values(by='Importance', ascending=False)

# In bảng
print(importance_df)

# Vẽ biểu đồ
plt.figure(figsize=(10,6))
plt.barh(importance_df['Feature'], importance_df['Importance'], color='skyblue')
plt.xlabel("Feature Importance")
plt.title("Tầm quan trọng của các biến (XGBoost)")
plt.gca().invert_yaxis()  # để biến quan trọng nhất ở trên cùng
plt.grid(axis='x')
plt.tight_layout()
plt.show()
```
