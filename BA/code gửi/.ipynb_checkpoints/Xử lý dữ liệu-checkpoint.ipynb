# Một số thư viện Python

Trong Python, Pandas và NumPy là hai thư viện cốt lõi được sử dụng rộng rãi trong xử lý dữ liệu, phân tích dữ liệu, và tính toán khoa học.
 ## Vai trò của NumPy
NumPy (Numerical Python) là thư viện nền tảng cho tính toán số học và xử lý mảng đa chiều trong Python. Nó cung cấp các công cụ mạnh mẽ để làm việc với dữ liệu số, đặc biệt trong các ứng dụng khoa học và kỹ thuật.

Mảng đa chiều (ndarray): NumPy cung cấp đối tượng mảng (ndarray) hiệu quả để lưu trữ và thao tác dữ liệu số (số nguyên, số thực, v.v.). Mảng này nhanh hơn nhiều so với danh sách Python nhờ tối ưu hóa ở cấp thấp.

Tính toán số học: Hỗ trợ các phép toán ma trận, đại số tuyến tính, thống kê cơ bản (trung bình, phương sai, v.v.), và các phép toán element-wise (theo từng phần tử).

Hiệu suất cao: Được viết bằng C, NumPy rất nhanh và tối ưu cho các phép toán trên tập dữ liệu lớn.

Hỗ trợ toán học: Cung cấp các hàm toán học như sin, cos, log, exp, và các công cụ cho phép tạo số ngẫu nhiên, biến đổi Fourier, v.v.

Nền tảng cho các thư viện khác: NumPy là cơ sở cho nhiều thư viện khác như Pandas, SciPy, TensorFlow, và Scikit-learn.

 Ứng dụng:
 
Xử lý dữ liệu số lớn (ví dụ: ma trận trong học máy).
Tính toán khoa học (vật lý, kỹ thuật).
Xử lý hình ảnh, tín hiệu, và dữ liệu đa chiều.

## Vai trò của Pandas
Pandas được xây dựng dựa trên NumPy và chuyên biệt cho việc xử lý và phân tích dữ liệu có cấu trúc (như bảng dữ liệu, giống Excel hoặc SQL). Nó cung cấp các cấu trúc dữ liệu linh hoạt và dễ sử dụng như Series và DataFrame.


Cấu trúc dữ liệu:

Series: Mảng một chiều có gắn nhãn (index), giống như một cột trong bảng.

DataFrame: Bảng dữ liệu hai chiều (hàng và cột), tương tự bảng Excel hoặc bảng trong cơ sở dữ liệu.

Xử lý dữ liệu:

Đọc/ghi dữ liệu từ nhiều nguồn như CSV, Excel, SQL, JSON, v.v.

Lọc, sắp xếp, nhóm, và gộp dữ liệu (merge, join).

Xử lý dữ liệu bị thiếu (missing data) và làm sạch dữ liệu.

Phân tích dữ liệu: Hỗ trợ các phép toán thống kê, tính toán theo nhóm (groupby), và xử lý dữ liệu thời gian (time series).

Dễ sử dụng: Cú pháp đơn giản, trực quan, phù hợp cho người làm việc với dữ liệu dạng bảng.

Ứng dụng:

Phân tích dữ liệu khám phá (EDA - Exploratory Data Analysis).

Tiền xử lý dữ liệu cho học máy.

Làm việc với dữ liệu lớn từ các nguồn như CSV, cơ sở dữ liệu, hoặc API.

Xử lý dữ liệu thời gian thực hoặc dữ liệu có cấu trúc phức tạp.

# Chuẩn hóa dữ liệu
1. Chuẩn hóa z-scre
   $$z= \frac{(x-\bar x)}{s}$$ 
$s$ là độ lệch chuẩn mẫu.

2. Chuẩn hóa min max
    $$z= \frac{(x-\min x)}{\max x - \min x}$$ 


```python
# chuẩn hóa dữ liệu
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['Credit_Limit', 'Avg_Open_To_Buy']] = scaler.fit_transform(df[['Credit_Limit', 'Avg_Open_To_Buy']])
```
